## Per module

```python
import pandas as pd

def per_module(df):
    # Extract first token from filename
    df["first_token"] = df["filename"].str.split("/").str[0]

    # Aggregate by first token
    agg_df = (
        df.groupby("first_token")[["loc_physical", "loc", "cc", "fan_in", "fan_out"]]
        .sum()
        .reset_index()
    )
    return agg_df
```

```python
df = pd.read_csv('linux_kernal.csv')
agg_df = per_module(df)
agg_df.to_csv('linux_kernal_agg.csv', index=False)
```

```python
df = pd.read_csv('pytorch.csv')
agg_df = per_module(df)
agg_df.to_csv('pytorch_agg.csv', index=False)
```

```python
df = pd.read_csv('kafka.csv')
agg_df = per_module(df)
agg_df.to_csv('kafka_agg.csv', index=False)
```
